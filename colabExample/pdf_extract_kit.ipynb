# Setup Config files

For each model that you need to run, you should change the .yaml file.

layout_detection.yaml:
```
inputs: /content/docs/ # This is src of input file
outputs: /content/outputs/layout_detection #output
tasks:
  layout_detection:
    model: layout_detection_yolo
    model_config:
      img_size: 1024
      conf_thres: 0.25
      iou_thres: 0.45
      model_path: /content/models/Layout/YOLO/ # This is important !
      doclayout_yolo_ft.pt
      visualize: True
```
another example,
ocr.yaml:

```
inputs: /content/docs/
outputs: /content/outputs/layout_detection
tasks:
  layout_detection:
    model: layout_detection_yolo
    model_config:
      img_size: 1024
      conf_thres: 0.25
      iou_thres: 0.45
      model_path: /content/models/Layout/YOLO/doclayout_yolo_ft.pt
      visualize: True
```

and the same for other .yaml files.
